```python
import numpy as np
import sentencepiece as spm

sp = spm.SentencePieceProcessor(model_file="bn_tokenizer.model")

def test_tokenize(corpus_path, n_stories=5000, out_file="test_train.bin"):
    buffer = []
    total_tokens = 0

    with open(corpus_path, "r", encoding="utf-8") as f:
        for i, line in enumerate(f):
            if i >= n_stories:
                break
            story = line.strip()
            if not story:
                continue
            ids = sp.encode(story, out_type=int)
            ids.append(sp.eos_id() if sp.eos_id() != -1 else 0)
            buffer.extend(ids)
            total_tokens += len(ids)

    arr = np.array(buffer, dtype=np.uint16)
    arr.tofile(out_file)
    print(f"Stories processed: {i+1}")
    print(f"Total tokens: {total_tokens:,}")
    print(f"File size: {arr.nbytes / 1e6:.2f} MB")
    return arr

test_arr = test_tokenize("bangla_corpus.txt", n_stories=5000)
```

```python
# Decode the first ~50 tokens back to text as a spot check
sample_ids = test_arr[:50].tolist()
print(sp.decode(sample_ids))
```

```python
import numpy as np
import sentencepiece as spm

sp = spm.SentencePieceProcessor(model_file="bn_tokenizer.model")
corpus_path = "bangla_corpus.txt"

# Pass 1: count total lines (fast, streams through the file without loading it)
with open(corpus_path, "r", encoding="utf-8") as f:
    total_lines = sum(1 for _ in f)

val_cutoff = int(total_lines * 0.99)  # last 1% held out for validation
print(f"Total stories: {total_lines:,} | train: {val_cutoff:,} | val: {total_lines - val_cutoff:,}")

# Pass 2: tokenize and flush to disk periodically
flush_every = 500_000  # tokens per flush, not stories

def run_full_tokenize():
    buffer_train, buffer_val = [], []
    total_train_tokens, total_val_tokens = 0, 0

    with open(corpus_path, "r", encoding="utf-8") as fin, \
         open("train.bin", "wb") as ftrain, \
         open("validation.bin", "wb") as fval:

        for i, line in enumerate(fin):
            story = line.strip()
            if not story:
                continue

            ids = sp.encode(story, out_type=int)
            ids.append(sp.eos_id() if sp.eos_id() != -1 else 0)

            if i < val_cutoff:
                buffer_train.extend(ids)
                total_train_tokens += len(ids)
                if len(buffer_train) >= flush_every:
                    np.array(buffer_train, dtype=np.uint16).tofile(ftrain)
                    buffer_train = []
            else:
                buffer_val.extend(ids)
                total_val_tokens += len(ids)
                if len(buffer_val) >= flush_every:
                    np.array(buffer_val, dtype=np.uint16).tofile(fval)
                    buffer_val = []

            if (i + 1) % 200_000 == 0:
                print(f"Processed {i+1:,} / {total_lines:,} stories...")

        # flush remainders
        if buffer_train:
            np.array(buffer_train, dtype=np.uint16).tofile(ftrain)
        if buffer_val:
            np.array(buffer_val, dtype=np.uint16).tofile(fval)

    print(f"Done. Train tokens: {total_train_tokens:,} | Val tokens: {total_val_tokens:,}")

run_full_tokenize()
```
